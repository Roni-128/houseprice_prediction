```python
import numpy as np
from sklearn.linear_model import LinearRegression
x = np.array([[1],[2],[3],[4]])
y = np.array([3,5,7,9])
model = LinearRegression()
model.fit(x,y)
pred = model.predict([[6]])
print("prediction : ",pred)
print("slope : ",model.coef_)
print("intercept : ",model.intercept_)
```

```python
from sklearn.linear_model import LinearRegression
x = np.array([[200],[300],[400],[500],[600]])
y = np.array([40000,90000,160000,250000,360000])
model = LinearRegression()
model.fit(x,y)
size = int(input("enter the house size : "))
print("the house size is : ",size)
prize = model.predict([[size]])
print("the predicted house prize : ",prize)
print("slope : ",model.coef_)
print("intercept : ",model.intercept_)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
x = np.array([[200],[300],[400],[500],[600]])
y = np.array([40000,90000,160000,250000,360000])
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=.4)
model = LinearRegression()
model.fit(x_train,y_train)
y_pred = model.predict(x_test)
score = r2_score(y_test,y_pred)
print("actual : ",y_test)
print("predicted : ",y_pred)
print("accuracy(r2 score ): ",score)
```
